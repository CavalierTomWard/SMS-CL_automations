```python
from google.colab import drive
import os
import glob
import pandas as pd

# 1. Mount Google Drive
drive.mount('/content/drive')

# 2. Define your backup folder path
folder_path = '/content/drive/MyDrive/SMSB&RP'

# 3. Automatically find the most recent CSV file (e.g., master stats or text-only SMS)
def get_latest_file(pattern):
    search_path = os.path.join(folder_path, pattern)
    files = glob.glob(search_path)
    if not files:
        return None
    # Sort files by modification time (newest first)
    latest_file = max(files, key=os.path.getmtime)
    return latest_file

# Target the master contact stats or your exported SMS CSV
latest_stats_csv = get_latest_file('master_contact_stats_*.csv')
latest_sms_csv = get_latest_file('sms_text_only_*.csv')

print(f"Targeting Master Stats: {latest_stats_csv}")
print(f"Targeting SMS Data: {latest_sms_csv}")

# 4. Example: Load the latest master stats and view top active contacts
if latest_stats_csv:
    df = pd.read_csv(latest_stats_csv)
    print("\nTop 5 Most Active Contacts:")
    print(df[['contact_name', 'phone_number', 'total_interactions', 'total_call_mins']].head(5))
else:
    print("No master stats CSV found yet. Run the merger script first.")
```
